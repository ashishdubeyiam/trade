# Exploratory Data Analysis (on Dummy Data)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_style('whitegrid')
```

```python
df = pd.read_csv('../data/dummy_fraud_dataset.csv')
print(df.head())
```

```python
print("--- DataFrame Info ---")
df.info()
print("\n--- Descriptive Statistics (including all columns) ---")
print(df.describe(include='all'))
print("\n--- Missing Values Check ---")
print(df.isnull().sum())
```

## Numerical Feature Analysis

```python
numerical_cols_to_plot = ['Transaction_Amount', 'Account_Balance', 'Risk_Score', 'Card_Age']

for col in numerical_cols_to_plot:
    if col in df.columns:
        plt.figure(figsize=(8, 5))
        sns.histplot(df[col], kde=True)
        plt.title(f'Histogram of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
        plt.show()
    else:
        print(f"Column '{col}' not found for histogram.")
```

```python
numerical_cols_to_plot = ['Transaction_Amount', 'Account_Balance', 'Risk_Score', 'Card_Age']

for col in numerical_cols_to_plot:
    if col in df.columns and 'Fraud_Label' in df.columns:
        plt.figure(figsize=(8, 5))
        sns.boxplot(x='Fraud_Label', y=col, data=df)
        plt.title(f'Box Plot of {col} by Fraud_Label')
        plt.xlabel('Fraud_Label')
        plt.ylabel(col)
        plt.show()
    else:
        print(f"Column '{col}' or 'Fraud_Label' not found for box plot.")
```

## Categorical Feature Analysis

```python
categorical_cols_to_plot = ['Transaction_Type', 'Device_Type', 'Location', 'Merchant_Category', 'Card_Type', 'Authentication_Method']

for col in categorical_cols_to_plot:
    if col in df.columns:
        plt.figure(figsize=(10, 6))
        sns.countplot(y=col, data=df, order = df[col].value_counts().index)
        plt.title(f'Count Plot of {col}')
        plt.xlabel('Count')
        plt.ylabel(col)
        plt.show()
    else:
        print(f"Column '{col}' not found for count plot.")
```

```python
categorical_cols_for_target_dist = ['Transaction_Type', 'Device_Type', 'Location']

for col in categorical_cols_for_target_dist:
    if col in df.columns and 'Fraud_Label' in df.columns:
        plt.figure(figsize=(10, 6))
        sns.countplot(x=col, hue='Fraud_Label', data=df, order = df[col].value_counts().index)
        plt.title(f'Distribution of Fraud_Label within {col}')
        plt.xlabel(col)
        plt.ylabel('Count')
        plt.xticks(rotation=45, ha='right')
        plt.legend(title='Fraud_Label')
        plt.show()
    else:
        print(f"Column '{col}' or 'Fraud_Label' not found for target distribution count plot.")
```

## Correlation Analysis

```python
numerical_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(12, 10))
sns.heatmap(numerical_df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
